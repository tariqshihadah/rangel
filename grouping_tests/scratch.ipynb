```python
size = 1000
l1 = np.arange(0, size, 1).reshape(-1, 1)
l2 = np.arange(0, size, 1).reshape(1, -1)
arr = np.arange(0, size**2, 1).reshape(size, size)
```

```python
%%timeit
x = np.minimum(l1, l2)
np.where(arr > x, x, arr)
```

```python
%%timeit
x = np.minimum(l1, l2)
x = np.minimum(arr, x)
```

```python
%%timeit
x = np.minimum(l1, l2)
np.minimum(arr, x, out=arr)
```
